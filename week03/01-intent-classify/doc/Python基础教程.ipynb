# 面向对象基础

类可以用来描述一类对象。对象可以拥有属性和方法。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def get_level(self):
        if self.score >= 90:
            return "优秀"
        elif self.score >= 80:
            return "良好"
        elif self.score >= 60:
            return "及格"
        else:
            return "不及格"

    def introduce(self):
        return f"我是 {self.name}，成绩是 {self.score}，等级是 {self.get_level()}"

student = Student("小明", 92)
print(student.introduce())
```

# 实操项目：学生成绩管理

下面实现一个简单的学生成绩管理系统，包含添加学生、查看学生、计算平均分、找最高分等功能。

```python
students = []

def add_student(name, score):
    student = {
        "name": name,
        "score": score
    }
    students.append(student)

def show_students():
    if not students:
        print("暂无学生数据")
        return

    for student in students:
        print(f"{student['name']}：{student['score']} 分")

def get_average_score():
    if not students:
        return 0

    total = sum(student["score"] for student in students)
    return total / len(students)

def get_top_student():
    if not students:
        return None

    return max(students, key=lambda student: student["score"])

add_student("小明", 92)
add_student("小红", 85)
add_student("小刚", 73)
add_student("小丽", 96)

show_students()
print("平均分：", get_average_score())
print("最高分学生：", get_top_student())
```

# 实操项目：文本词频统计

这个项目会读取一段文本，统计每个词出现的次数。

```python
text = '''
python is useful
python is easy
data is powerful
python and data are important
'''

words = text.lower().split()
word_count = {}

for word in words:
    if word not in word_count:
        word_count[word] = 1
    else:
        word_count[word] += 1

print(word_count)

sorted_word_count = sorted(word_count.items(), key=lambda item: item[1], reverse=True)

for word, count in sorted_word_count:
    print(word, count)
```

# 综合练习

下面完成一个商品销售额统计任务：创建商品列表，计算每个商品的销售额，找出销售额最高的商品，统计总销售额，并把结果写入文本文件。

```python
products = [
    {"name": "键盘", "price": 199, "sales": 30},
    {"name": "鼠标", "price": 59, "sales": 120},
    {"name": "显示器", "price": 899, "sales": 15},
    {"name": "耳机", "price": 299, "sales": 45}
]

for product in products:
    product["revenue"] = product["price"] * product["sales"]

top_product = max(products, key=lambda product: product["revenue"])
total_revenue = sum(product["revenue"] for product in products)

print("商品数据：")
for product in products:
    print(product)

print("销售额最高的商品：", top_product)
print("总销售额：", total_revenue)

with open("sales_report.txt", "w", encoding="utf-8") as f:
    f.write("销售报告\n")
    for product in products:
        f.write(f"{product['name']}：销售额 {product['revenue']}\n")
    f.write(f"总销售额：{total_revenue}\n")

print("销售报告已写入 sales_report.txt")
```
